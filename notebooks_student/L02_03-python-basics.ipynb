### Nested if-else

```python
# nested if-else example
customers = {"Tom": {"mail": True, "phone": False},
             "Julie": {"mail": False, "phone": False},
             "Jasper": {"mail": True, "phone": True}}

contact_customer = "tom"

# check if the customer we want to contact is in our customer list
if contact_customer in customers.keys():
    
    # check if customer wants to be contacted via mail
    if customers[contact_customer]["mail"] == True:
        print("customer wants to be contacted via mail")  
    else:
        print("customer wants not to be contacted via mail")
        
    # check if customer wants to be contacted via phone
    if customers[contact_customer]["phone"] == True:
        print("customer wants to be contacted via phone")  
    else:
        print("customer wants not to be contacted via phone")
        
else:
    
    print("customer not in customer list!")
```

## Try-Except

### Definition

```python
customer_spendings = {"Tom": {"years_customer": 5, "total_spendings": 304},
                  "Julie": {"years_customer": 0, "total_spendings": 1402},
                  "Jack": {"years_customer": 2, "total_spendings": 459}}

for customer, value in customer_spendings.items():
    
    try:
        avg_spending_per_year = value["total_spendings"] / value["years_customer"]
        print(customer, ":", avg_spending_per_year)
    except:
        avg_spending_per_year = value["total_spendings"]
        print(customer, ":", avg_spending_per_year, "for less than 1 year a customer")
```

### Return

```python
# define function with multiple return statements
def divide_and_sum(number, constant_1, constant_2):
    
    # divide number by constant_1
    result_1 = number / constant_1
    return(result_1)

    # add constant_2 to result_1
    result_2 = result_1 + constant_2
    return(result_2)
```

```python
# call function 
result = divide_and_sum(number=10, constant_1=5, constant_2=2)
print(result)
```

### Flexible Functions

```python
def print_args(*args):
   # loop through non-keyword arguments (stored as a tuple)
   for arg in args:
      print(arg)
      
print_args(1, 2, 3)
```

```python
def print_kwargs(**kwargs):
   # loop through keyword arguments (stored as a dictionary)
   for key, value in kwargs.items():
      print("{key}: {value}".format(key=key, value=value))

print_kwargs(a=1, b=2, c=3)
```

```python
def print_args_kwargs(*args, **kwargs):
   # loop through non-keyword arguments (stored as a tuple)
   for arg in args:
      print("Non-keyword argument: {arg}".format(arg=arg))
            
   # loop through keyword arguments (stored as a dictionary)
   for key, value in kwargs.items():
      print("Keyword argument: {key}={value}".format(key=key, value=value))

print_args_kwargs(1, 2, 3, a=1, b=2, c=3)
```
