# Installation Guide {.appendix}

This appendix walks through setting up Python and an IDE (VS Code is recommended) on your own
laptop for this course, so that every chapter in this book runs exactly as shown. The idea is
to create a dedicated, self-contained **virtual environment** with a known-good set of package
versions, rather than relying on whatever happens to already be installed on your machine.


## Getting the course files


Before installing anything, get a local copy of the book's repository. The notebooks refer to
`requirements.txt` and the `data/` folder by a relative path, so those only resolve correctly
once you have the whole project folder, not just one downloaded notebook. Downloading a single
notebook by itself, for example via right-click "Save link as" on one file, will not work: it
won't come with `requirements.txt` or `data/` alongside it, and every `pip install` or
`pd.read_csv` call in this book assumes they're there.

The simplest way, no `git` required:

1. Go to the book's [GitHub repository](https://github.com/MatthBogaert/IDAFBE-book).
2. Click the green **Code** button, then **Download ZIP**.
3. Unzip it somewhere on your machine. This folder is your project root for the rest of this
   appendix and for every notebook in the course.

If you're already comfortable with `git`, cloning works the same way and makes it easier to
pull future updates to the course material:

```bash
git clone https://github.com/MatthBogaert/IDAFBE-book.git
```


## Create a virtual environment


A virtual environment is a self-contained directory that holds a Python installation for a
particular version of Python, plus a specific set of additional packages. Without one, packages
installed for different projects can silently conflict with each other: a version of `pandas`
that one project needs might break another. Learning to use virtual environments now is a
worthwhile habit: it's exactly how you'll be expected to manage Python projects later, e.g. in a
company.

In this book `conda` is used to create the environment (you should already have `conda` available if you
installed Anaconda). This book pins everything to **Python 3.11**: see
`requirements.txt` in the project root for the exact package versions.


### VS Code


If you'd like a visual walkthrough of creating a conda environment from within VS Code, this
[video](https://www.youtube.com/watch?v=fnb4_MzpZFU) covers the general steps. Note that the
video uses its own example environment name and Python version: when it gets to
`conda create --name mynewenv`, use the command from the next section instead, with **this
book's** environment name and Python version.


### Command line


Open a terminal (not a notebook cell) and run:

```bash
conda create --name DA2026 python=3.11
```

Then activate it:

```bash
conda activate DA2026
```

You may need to restart VS Code (or your notebook) afterward and explicitly select this new
environment as your kernel: see [Selecting the right kernel](#selecting-the-right-kernel)
below.


## Installing required packages


With the `DA2026` environment activated, install the exact package versions this book was


## Selecting the right kernel


Installing the packages into the `DA2026` environment isn't enough on its own: VS Code (or
Jupyter) also needs to actually *use* that environment when it runs your code. This trips
people up more often than the installation itself, so it's worth being deliberate about it.

**For `.ipynb` notebooks:** open the notebook, click the kernel name in the top-right corner of
the notebook editor, choose **"Select Another Kernel..."**, and pick **"Python (DA2026)"** from
the list.

**For your overall Python interpreter** (affects linting, IntelliSense, and some extensions):
`Cmd+Shift+P` (or `Ctrl+Shift+P` on Windows/Linux) → **"Python: Select Interpreter"** → choose
the interpreter inside your `DA2026` environment.

**For `quarto preview`/`quarto render` on `.qmd` files:** this project's `_quarto.yml` already
pins `jupyter: da2026` at the top, so Quarto picks the right kernel automatically regardless of
whatever interpreter VS Code itself is set to. If a preview still seems to be using the wrong
environment, close and reopen the preview tab: it can cache the kernel it started with from
before you had `DA2026` set up.
