## Testing the relationships 
Let's now take a closer look at the **relationship** between the customers and subscriptions tables.

A key question to explore is: *How many subscriptions does each customer have?*. This will help us understand the structure of the data and inform what an Entity-Relationship Diagram (ERD) might look like. To do so, we should do a left outer join on the customer and subscriptions table and count the number of subscriptions per customers.

We expect that a single customer can have multiple subscriptions, indicating a one-to-many relationship from customers to subscriptions.

```python
# Merge customers with subscriptions to include all customers (even those with zero subscriptions)
merged = pd.merge(
    customers[["CustomerID"]],
    subscriptions[["CustomerID", "SubscriptionID"]],
    on="CustomerID",
    how="left"
)

# Count the number of subscriptions per customer
agg = merged.groupby("CustomerID")["SubscriptionID"].count().reset_index() # reset_index() is needed to keep both customerID and SubscriptionID as columns
agg = agg.rename(columns={"SubscriptionID": "NumSubscriptions"})

# Show summary statistics
print("Min:", agg["NumSubscriptions"].min(), "Max:", agg["NumSubscriptions"].max())
```

## Working with dates

Aas you might have noticed, we have kept the dated untouched until now. They are by default read-in as `object` or `string` and converted afterwards in pandas to datetime unsing `pd.to_datetime()`, which automatically handles most formats and can be customized with a `format` argument if needed.

Some of the most important date format codes in Python are:
- `%Y`: 4-digit year (e.g., 2025)  
- `%y`: 2-digit year (e.g., 25 for 2025)  
- `%m`: 2-digit month (01–12)  
- `%d`: 2-digit day of month (01–31)  
- `%H`: Hour (00–23, 24-hour clock)  
- `%I`: Hour (01–12, 12-hour clock)  
- `%p`: AM/PM  
- `%M`: Minute (00–59)  
- `%S`: Second (00–59)

Let's first change the dates in the **subscriptions** first and then the **customers** table. 

```python
# Select the date columns in subscriptions DataFrame
date_cols = ["StartDate", "EndDate", "PaymentDate", "RenewalDate"]

# Dates are in "day/month/year" format, specify the format
# Instead of looping, we will use apply. It is less structured but more efficient, especially for large datasets with many columns.
subscriptions[date_cols] = subscriptions[date_cols].apply(
    lambda col: pd.to_datetime(col, format="%d/%m/%Y", errors="coerce") #errors="coerce" will set invalid parsing to NaT (missing date)
    )

# The for loop would look like this:
#for col in date_cols:
#    subscriptions[col] = pd.to_datetime(subscriptions[col], format="%d/%m/%Y", errors="coerce")

# Check the result
print(subscriptions[date_cols].dtypes)
subscriptions[date_cols].head()
```

```python
# For the customers table
customers["DOB"] = pd.to_datetime(customers["DOB"], format="%d/%m/%Y", errors="coerce")

# Check the result
print(customers["DOB"].dtypes)
customers["DOB"].head()
```
