## Create a virtual environment

It is recommended to create a virtual environment to install the required packages.
Essentially, a virtual environment is a self-contained directory
that contains a Python installation for a particular version of Python, plus a number of additional packages.
Not using a virtual environment can lead to conflicts between packages and versions.

> **TIP:**
> Creating virtual environments is a best practice when working with Python projects.
> It is a useful skill to learn now, since you will use it in the future when you go work inside a company.

We will use conda to create the virtual environment
(you should have conda installed when you installed Anaconda for your jupyter notebook). We will install Python 3.11, since this is the version this course is pinned to (see `requirements.txt`). When creating new Python environments in general, it is advised to use one of the most recent stable versions.

### Command Line

To create a virtual environment, run the following command in a bash terminal window (not in a notebook window!):

```bash
conda create --name DA2026 python=3.11
```

To activate the virtual environment, run the following command:

```bash
conda activate DA2026
```


It might be required to restart your notebook (or VS Code) and select the correct kernel afterward — see **Selecting the right kernel** below.

## Installing required packages

> **TIP:**
> Make sure that the `requirements.txt` file is in the project root, alongside this notebook.

With the `DA2026` environment activated, you can install the exact package versions this course was written and tested against by running the following command
(this might take a while).

```bash
pip install -r requirements.txt
```

These versions are pinned deliberately — for example, this course stays on `pandas>=2.3,<3` rather than the newest `pandas` release, since a few packages we rely on haven't been fully verified against `pandas` 3.0 yet. Installing anything other than the pinned `requirements.txt` is the most common reason code that works in the lectures doesn't work on your own machine.

## Selecting the right kernel

Installing the packages into the `DA2026` environment isn't enough on its own — VS Code (or Jupyter) also needs to actually *use* that environment when it runs your code. This trips people up more often than the installation itself.

**For this notebook:** click the kernel name in the top-right corner of the notebook editor, choose **"Select Another Kernel..."**, and pick **"Python (DA2026)"** from the list.

**For your overall Python interpreter** (affects linting, IntelliSense, and some extensions): `Cmd+Shift+P` (or `Ctrl+Shift+P` on Windows/Linux) → **"Python: Select Interpreter"** → choose the interpreter inside your `DA2026` environment.

> **IMPORTANT:**
> If you have multiple conda environments with the *same* Python version, they can look identical in VS Code's picker unless you check the environment name specifically. Picking the wrong one usually shows up as a confusing error like `AttributeError: _ARRAY_API not found` or `A module that was compiled using NumPy 1.x cannot be run in NumPy 2.x` — that specific error means the active kernel isn't actually `DA2026`. Re-check the kernel/interpreter selection above.
